```python
from bs4 import BeautifulSoup
import pandas as pd 
import requests # Biblioteca para consumo de API
import time 
```

```python
url = (r"https://lista.mercadolivre.com.br/acessorios-veiculos/pecas-carros-caminhonetes/suspensao-direcao/amortecedores/amortecedor-polo")

```

```python
#Cabeçalho HTTP para autenticação WEB, (sem o headers o servidor identifica a automação)
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
    "Accept-Language": "pt-BR,pt;q=0.9,en-US;q=0.8,en;q=0.7","Referer": "https://www.mercadolivre.com.br/"}
#Metodo de requisição web
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

print (f"Status code:{response.status_code}")
```

```python
#Transforma o HTML da página  url em um objeto BeautifulSoup 
soup = BeautifulSoup(response.text, "html.parser")
#Localiza os elementos <li> que representam os cards dos produtos no Mercado livre
produtos = soup.find_all("li", class_= "ui-search-layout__item")
#Verifica quantos anúncios foram capturados na primeira página
print ("Titulo da página recebida:", soup.title.string if soup.title else "Sem título")

titulos_h2 = soup.find_all("h2")
print(f"Total de tags <h2> encontradas: {len(titulos_h2)}")

if titulos_h2: 
    print("Exemplo de classe do primeiro H2:", titulos_h2[0].get("class"))
    print("Exemplo de texto do primeiro H2:", titulos_h2[0].text.strip())
```
