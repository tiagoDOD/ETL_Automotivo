```python
url = "https://api.mercadolivre.com/sites/MLB/search?q=amortecedor-polo"
#Cabeçalho HTTP para simulando uma pesquisa no navegador
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8",
    "Accept-Language": "pt-BR,pt;q=0.9,en-US;q=0.8,en;q=0.7",
    "Referer": "https://www.mercadolivre.com.br/"}
#Chamada de requisição da URL com o formato headers para "burlar" o reconhecimento de automação
response = requests.get(url, headers= headers) 
```
